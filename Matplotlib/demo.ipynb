```python
#Pyplot API
data={
    "salary":[25000,50000,40000,60000,30000,45000,20000,70000,65000]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
#Line plot
plt.plot(df["salary"],color="orange",marker='o',linestyle='dashed',linewidth=2,markersize=10)
plt.show()
```

```python
#Histogram
plt.hist(df["salary"],bins=5,color="green",edgecolor="black")
plt.show()
```

```python
#Boxplot
plt.boxplot(df["salary"])
```

```python
df["dept"]=["HR","IT","Finance","HR","HR","Sales","IT","Finance","Sales"]
```

```python
#Pie chart
count=df["dept"].value_counts()
print(count)

plt.pie(count,labels=count.index,autopct="%1.1f",explode=[0,0.1,0.2,0])
```

```python
#Count plot
plt.bar(count.index,count,color=["red","green","blue","yellow"])
```

```python
df["Age"]=[25,30,28,35,32,29,31,27,26]
print(df)
```

```python
#scatter plot
plt.scatter(df["Age"],df["salary"],color="orange",marker='o')
plt.show()
```

```python
sort_age=df.sort_values(by="Age")
print(sort_age)
```

```python
#line plot
plt.plot(sort_age["Age"],sort_age["salary"],color="green",marker='o',linestyle='dashed',linewidth=2,markersize=10)
plt.show()
```

```python
#bar plot
plt.bar(sort_age["Age"],sort_age["salary"],color="blue",edgecolor="black")
plt.show()
```

```python
#bivariate variable analysis

hr_sal=df[df["dept"]=="HR"]["salary"]
hr_sal
```

```python
fin_sal=df[df["dept"]=="Finance"]["salary"]
fin_sal
```

```python
sales_sal=df[df["dept"]=="Sales"]["salary"]
sales_sal
```

```python
it_sal=df[df["dept"]=="IT"]["salary"]
it_sal
```

```python
#boxplot

plt.boxplot([hr_sal,fin_sal,sales_sal,it_sal])
```
