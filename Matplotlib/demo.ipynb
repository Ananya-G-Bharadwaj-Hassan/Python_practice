```python
#Pyplot API
data={
    "salary":[25000,50000,40000,60000,30000,45000,20000,70000,65000]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
#Line plot
plt.plot(df["salary"],color="orange",marker='o',linestyle='dashed',linewidth=2,markersize=10)
plt.show()
```

```python
#Histogram
plt.hist(df["salary"],bins=5,color="green",edgecolor="black")
plt.show()
```

```python
#Boxplot
plt.boxplot(df["salary"])
```

```python
df["dept"]=["HR","IT","Finance","HR","HR","Sales","IT","Finance","Sales"]
```

```python
#Pie chart
count=df["dept"].value_counts()
print(count)

plt.pie(count,labels=count.index,autopct="%1.1f",explode=[0,0.1,0.2,0])
```

```python
#Count plot
plt.bar(count.index,count,color=["red","green","blue","yellow"])
```

```python
df["Age"]=[25,30,28,35,32,29,31,27,26]
print(df)
```

```python
#scatter plot
plt.scatter(df["Age"],df["salary"],color="orange",marker='o')
plt.show()
```

```python
sort_age=df.sort_values(by="Age")
print(sort_age)
```

```python
#line plot
plt.plot(sort_age["Age"],sort_age["salary"],color="green",marker='o',linestyle='dashed',linewidth=2,markersize=10)
plt.show()
```

```python
#bar plot
plt.bar(sort_age["Age"],sort_age["salary"],color="blue",edgecolor="black")
plt.show()
```

```python
#bivariate variable analysis

hr_sal=df[df["dept"]=="HR"]["salary"]
hr_sal
```

```python
fin_sal=df[df["dept"]=="Finance"]["salary"]
fin_sal
```

```python
sales_sal=df[df["dept"]=="Sales"]["salary"]
sales_sal
```

```python
it_sal=df[df["dept"]=="IT"]["salary"]
it_sal
```

```python
#boxplot

plt.boxplot([hr_sal,fin_sal,sales_sal,it_sal])
```

```python
#pie chart

salary_by_dept=df.groupby("dept")["salary"].sum()
salary_by_dept
plt.pie(salary_by_dept,labels=salary_by_dept.index,autopct="%1.1f")
```

```python
#bar plot

hr_mean=sum(hr_sal)/len(hr_sal)
fin_mean=sum(fin_sal)/len(fin_sal)
sales_mean=sum(sales_sal)/len(sales_sal)
it_mean=sum(it_sal)/len(it_sal)

plt.bar(["HR","Finance","Sales","IT"],[hr_mean,fin_mean,sales_mean,it_mean],color=["blue","green"],edgecolor="black")
plt.show()
```

```python
#multivariate variable analysis

df["experience"]=[2,5,3,7,4,6,1,8,9]
print(df)
```

```python
#bubble plot

plt.scatter(df["Age"],df["salary"],s=df["experience"]*50,color="orange",alpha=0.5)
plt.show()
```

```python
#Object Oriented API

fig, axs= plt.subplots(1,3,figsize=(15,5))

#Line plot

axs[0].plot(sort_age["Age"],sort_age["salary"],color="green",marker='o',linestyle='dashed',linewidth=2,markersize=10)
axs[0].set_title("Line plot")
axs[0].grid()
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

#Histogram

axs[1].hist(df["salary"],bins=5,color="green",edgecolor="black")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

#Boxplot

axs[2].boxplot(df["salary"])
axs[2].set_title("Boxplot")
axs[2].set_xlabel("Salary")
axs[2].set_ylabel("Values")
```

```python
#Multiple plot

salesdata={
    "Year":[2020,2021,2022,2023],
    "Sales":[1000,1500,2000,2500],
    "Expenses":[500,600,700,800],
    "Profit":[500,900,1300,1700]
}
```

```python
df=pd.DataFrame(salesdata)

plt.plot(df["Year"],df["Sales"],label="Sales")
plt.plot(df["Year"],df["Expenses"],label="Expenses")
plt.plot(df["Year"],df["Profit"],label="Profit")
plt.title("Sales, Expenses and Profit over Years")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
#3D plot

ax=plt.axes(projection='3d')
ax.scatter(df["Year"],df["Sales"],df["Expenses"],c='r',marker='o')
ax.set_xlabel("Year")
ax.set_ylabel("Sales")
ax.set_zlabel("Expenses")
```
