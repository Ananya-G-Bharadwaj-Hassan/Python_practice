```python
data=torch.tensor([
    [10,11,12,13],
    [20,21,22,23],
    [30,31,32,33]   
])
indices_to_select=torch.tensor([[2],[0],[3]])
selected_students=torch.gather(data,dim=1,index=indices_to_select)
print(selected_students)
```

```python
#forward pass
N=10
D_in=1
D_out=1
x=torch.randn(N,D_in)
```

```python
true_w=torch.tensor([[2.0]])
true_b=torch.tensor(1.0)
print(true_w, true_b)
y_true=x@true_w+true_b+torch.randn(N,D_out)*0.1
```

```python
w=torch.randn(D_in,D_out,requires_grad=True)
b=torch.randn(1,requires_grad=True)
print(w,b)
y_hat=x@w+b
print(y_hat)
```

```python
#backward pass
error=y_hat-y_true
mse=error**2
loss=mse.mean()
print(loss)
```

```python
loss.backward()

# print(f"Gradient for w: {w.grad}")
# print(f"Gradient for b: {b.grad}")
```

```python
#update parameters
with torch.no_grad():
    learning_rate=0.01
    w-=learning_rate*w.grad
    b-=learning_rate*b.grad
    w.grad.zero_()
    b.grad.zero_()
    print(w,b)
    epochs=100
for epoch in range(epochs):
    y_hat=x@w+b
    error=y_hat-y_true
    mse=error**2
    loss=mse.mean()
    loss.backward()
if loss.item()<0.01:
        print("Training complete")
        print(f"Final loss: {loss.item()}")
print(f"Final weights: {w}")
print(f"Final bias: {b}")
```
