```python
students=torch.tensor([[45.,40.,30.],[4.,5.,6.]])
print(students)
```

```python
avg=students.mean()
print(avg)
avg1=students.mean(dim=0)
print(avg)
avg2=students.mean(dim=1)
print(avg2)
```

```python
best_student=students.argmax(dim=1)
print(best_student)
```

```python
data=torch.tensor([
    [10,11,12,13],
    [20,21,22,23],
    [30,31,32,33]   
])
indices_to_select=torch.tensor([[2],[0],[3]])
selected_students=torch.gather(data,dim=1,index=indices_to_select)
print(selected_students)
```
